```python
while True:
    try:
        n = int(input("Введите N: "))
        if n >= 2:
            break
        print("N должно быть не меньше 2")
    except ValueError:
        print("Нужно ввести целое число")

numbers = list(range(2, n + 1))

i = 0
while i < len(numbers) and numbers[i] * numbers[i] <= n:
    p = numbers[i]
    for j in range(p * p, n + 1, p):
        if j in numbers:
            numbers.remove(j)
    i += 1

for p in numbers:
    print(p)
```
